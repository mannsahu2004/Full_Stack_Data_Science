```python
# 16Q. Write a program that checks if the input number is a prime number
num = int(input("Enter a number: "))
if num <= 1:
    print("Not prime")
else:
    for i in range(2,num):
        if num%i == 0:
            print("not prime")
            break
    else:
        print("prime")
```

```python
# 17Q.How can you accept a boolean value (True/False) from the user?
user_input = input("Enter True or False: ")

if user_input == "True":
    value = True
elif user_input == "False":
    value = False
else:
    print("Invalid input")
    value = None

print(value)
```
