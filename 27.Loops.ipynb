While Loop

```python
i = 1

while i<=5:
    print("Data Science: ", i)
    i = i + 1
```

```python
i = 5
while i>=1:
    print("Data Science: ",i)
    i = i - 1
```

```python
i = 1

while i <= 5:
    print("Data Science: ", i)
    j = 1
    while j <= 4:
        print("Technology: ", i)
        j = j + 1

    i = i + 1
    print()
```

```python
i = 1

while i <= 5:
    print("Data science", end = " ") # when we mention end then new line will not create
    j = 1
    while j <= 4:
        print("Technology", end = " ")
        j = j + 1

    i = i + 1
    print()
```

```python
# Lets use while loop using some numbers
i = 1

while i <= 2:
    j = 0
    while j <= 2:
        print(i*j, end = " ")
        j += 1
    print()
    i += 1
```

```python
i = 1

while i <= 4:
    j = 0
    while j <= 3:
        print(i*j, end = " ")
        j += 1
    print()
    i += 1
```

For Loop

```python
name = 'nit'

for i in name:
    print(i)
```

```python
name = [1, 3.5, 'hello']
for i in name:
    print(i)
```

```python
for i in [2, 3, 7.8, 'hi']:
    print(i)
```

```python
for i in range(1,51):
    if i%5 == 0:
        print(i)
```

```python
# Print the number which is divisible by 3
for i in range (1,31):
    if i%3 == 0:
        print(i)
```

```python
# Print the number which is not divisible by 3
for i in range(1,31):
    if i%3 != 0:
        print(i)
```

```python
for i in range(1, 51):
    if i%5 != 0:
        print(i)
```

Break

```python
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```
