Matrix Operations in NumPy

1. Element-wise Addition, Subtraction, and Division

```python
import numpy as np
import math

x = np.array([[1,2], [3,4]])
y = np.array([[5,6], [7,8]])

print("Addition:\n", np.add(x,y))
print("Subtraction:\n", np.subtract(x,y))
print("Division:\n", np.divide(x,y))
```

```python
x = np.array([[1,2], [3,4]])
y = np.array([[5,6], [7,8]])

print("The element wise addition of matrix is:\n",np.add(x,y))
print("The elnet wise subtraction of matrix is:\n",np.subtract(x,y))
print("The element wise division of matrix is:\n",np.divide(x,y))
```

2. Element-wise Multiplication vs. Matrix Multiplication

```python
#dot() : - Multipy elements of a row from first array with elements of a colums from the second array, and then add together .
import numpy as np

x = np.array([[1,2], [4,5]])
y = np.array([[7,8], [9,10]])

print("Element-wise multiplication:\n", np.multiply(x,y))
print("Matrix multiplication:\n", np.dot(x,y))
```

```python
import numpy as np
 
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.mod(a, b)
print(res)
```

Other Useful NumPy Matrix Functions

```python
#axis0 = columns
#axis1 = rows
#Transpose means converting rows into columns and columns into rows.

x = np.array([[1,2], [4,5]])
y = np.array([[7,8], [9,10]])

print("Square root:\n", np.sqrt(x))
print("Sum of all elements:\n", np.sum(y))

print("Column wise sum:", np.sum(y, axis=0))
print("Row wise sum:", np.sum(y, axis=1))
print("Transpose:\n", x.T)
```

Products of Vectors and Matrices in NumPy

```python
a = np.array([2,6])
b = np.array([3,10])

print("Inner product of vectors a and b =")
print(np.inner(a,b))

x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])

print("Inner product of vectors x and y =")
print(np.inner(x,y))
```

```python
a = np.array([2,6])
b = np.array([3,10])

print("Outer product of vectors a and b =")
print(np.outer(a,b))

x = np.array([[3, 6, 4], [9, 4, 6]])
y = np.array([[1, 15, 7], [3, 10, 8]])

print("Outer product of matrices x and y =")
print(np.outer(x,y))
```

Cross Product

```python
import numpy as np
#(a2​b3​−a3​b2​,a3​b1​−a1​b3​,a1​b2​−a2​b1​)
# Define vectors as 3D (z=0 for 2D compatibility)
a = np.array([3, 6, 0])
b = np.array([9, 10, 0])

print("Cross product of vectors a and b =")
print(np.cross(a, b))

# Define matrices (already 3D rows)
x = np.array([[2, 6, 9], [2, 7, 3]])
y = np.array([[7, 5, 6], [3, 12, 3]])

print("Cross product of matrices x and y =")
print(np.cross(x, y))
```

Stacking

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
res = np.stack((a,b), axis = 0)
print(res)
```

```python
#This example shows how stacking the same 1D arrays along axis 0, 1, and -1 changes the output shape.
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.stack((a,b), axis = 0))
print(np.stack((a,b), axis = 1))
print(np.stack((a,b), axis = -1))
```

```python
#This example stacks two 2D arrays along axis 0, 1, and 2 to show how the new 3D structure changes.
x = np.array([[1, 2, 3],
              [4, 5, 6]])

y = np.array([[7, 8, 9],
              [10, 11, 12]])

print(np.stack((x,y), axis=0))
print(np.stack((x,y), axis=1))
print(np.stack((x,y), axis=2))
```

```python
#This example stacks two 3D arrays along axis 0, 1, 2, and 3 to demonstrate how stacking works with higher-dimension data.
import numpy as np

m = np.array([[[1, 2], [3, 4]],
              [[5, 6], [7, 8]]])

n = np.array([[[10, 20], [30, 40]],
              [[50, 60], [70, 80]]])

print(np.stack((m, n), axis=0))
print(np.stack((m, n), axis=1))
print(np.stack((m, n), axis=2))
print(np.stack((m, n), axis=3))
```
